## 1) List / Dict / Set Comprehensions

Comprehensions are **compact ways** to build containers from iterables.

- List: `[expr for x in iterable if condition]`
- Set: `{expr for x in iterable if condition}`
- Dict: `{key_expr: val_expr for x in iterable if condition}`

They are equivalent to writing a `for` loop that appends/updates, but are often **clearer and faster** for simple transformations.



### 1.1 Transform: squares of the first 10 integers

Strings are immutable sequences of characters. Important tools:

### 2.1 Normalise names

**Task:** Given `names`, produce a **new list** where each name is stripped of spaces and title‑cased (`"joHN  doE "` → `"John Doe"`). Use a list comprehension + string methods.


```python

# YOUR TURN
names = ["   joHN  doE ", "ALICE   ", "  bOb SMITH"]
clean = [" ".join(part for part in name.strip().split()).title() for name in names]
clean
```


### 2.2 Palindrome test (ignoring punctuation and spaces)

**Task:** Write a function `is_palindrome(s)` that returns `True` if `s` is a palindrome **ignoring case, spaces and punctuation**.

> Tip: normalise with `lower()`, keep only `str.isalnum()` characters, then compare with its reverse `s[::-1]`.


```python

# YOUR TURN
def is_palindrome(s: str) -> bool:
    norm = "".join(ch.lower() for ch in s if ch.isalnum())
    return norm == norm[::-1]

tests = ["A man, a plan, a canal: Panama!", "racecar", "OpenAI", "Socorram-me, Subi no Onibus em Marrocos"]
{k: is_palindrome(k) for k in tests}
```


### 2.3 Word count dictionary

**Task:** Implement `word_count(s)` that returns a dict mapping each **lowercased, punctuation‑stripped** word to its frequency.


```python

# YOUR TURN
def word_count(s: str) -> dict[str, int]:
    words = [w.strip(".,;:!?\"'()[]").lower() for w in s.split()]
    words = [w for w in words if w]  # drop empties
    return {w: words.count(w) for w in set(words)}

text = "To be, or not to be: that is the question."
word_count(text)
```


### 2.4 Anagram checker

**Task:** Two strings are anagrams if they contain the same letters with the same multiplicities (ignoring spaces/punctuation/case).  
Write `are_anagrams(a, b)` returning `True/False`.

> Hint: normalise like in the palindrome, then compare **sorted strings** or **frequency dicts**.


```python

# YOUR TURN
def normalise_letters(s: str) -> str:
    return "".join(ch.lower() for ch in s if ch.isalpha())

def are_anagrams(a: str, b: str) -> bool:
    return sorted(normalise_letters(a)) == sorted(normalise_letters(b))

tests = [("Listen", "Silent"), ("Clint Eastwood", "Old West Action"), ("Apple", "Papel"), ("Hello", "Olelh!")]
{k: are_anagrams(*k) for k in tests}
```


### 2.5 Caesar cipher (encode/decode)

**Task:** Implement a simple Caesar cipher over letters `a..z` (preserve case; leave non‑letters unchanged).  
Write `caesar(s, k)` that shifts by `k` to **encode**, and `caesar(s, -k)` to decode.


```python

# YOUR TURN
def shift_char(ch: str, k: int) -> str:
    if ch.isalpha():
        base = ord('A') if ch.isupper() else ord('a')
        return chr(base + (ord(ch) - base + k) % 26)
    return ch

def caesar(s: str, k: int) -> str:
    return "".join(shift_char(ch, k) for ch in s)

msg = "Attack at Dawn!"
enc = caesar(msg, 3)
dec = caesar(enc, -3)
enc, dec
```

### 3.1 Sets & membership performance

**Task:** Compare membership checks in a **list** vs a **set** for large `n`.  
You don't need micro‑precision; just get a sense of the difference.

> Reminder: `x in list` is \(O(n)\); `x in set` is \(O(1)\) on average.


```python

# WARNING: choose n to fit your machine's memory/time.
import time

n = 2_000_000
target = n - 1

L = list(range(n))
S = set(range(n))

t0 = time.time(); _ = (target in L); t1 = time.time()
t2 = time.time(); _ = (target in S); t3 = time.time()

print(f"List membership:  {t1 - t0:.4f} s")
print(f"Set  membership:  {t3 - t2:.6f} s")
```


### 3.4 `while` loop bonus — double six

**Task:** Simulate rolls of two dice until the first **double six (6,6)** occurs. Return the number of rolls.  
Repeat the experiment 1000 times and estimate the mean.

> Hint: Use `while True:` and `return` when (6,6) appears.


```python

# YOUR TURN
import random

def rolls_until_double_six() -> int:
    rolls = 0
    while True:
        rolls += 1
        d1 = random.randint(1, 6)
        d2 = random.randint(1, 6)
        if d1 == 6 and d2 == 6:
            return rolls

N = 1000
avg = sum(rolls_until_double_six() for _ in range(N)) / N
avg
```

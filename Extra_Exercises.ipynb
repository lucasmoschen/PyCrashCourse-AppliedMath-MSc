## 1) List / Dict / Set Comprehensions

Comprehensions are **compact ways** to build containers from iterables.

- List: `[expr for x in iterable if condition]`
- Set: `{expr for x in iterable if condition}`
- Dict: `{key_expr: val_expr for x in iterable if condition}`

They are equivalent to writing a `for` loop that appends/updates, but are often **clearer and faster** for simple transformations.
